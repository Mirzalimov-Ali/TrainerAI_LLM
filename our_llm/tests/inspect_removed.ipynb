```python
from pathlib import Path
import json

PROJECT_ROOT = Path.cwd().parent.parent

RAW_DIR = PROJECT_ROOT / "data" / "raw"
FILTERED_DIR = PROJECT_ROOT / "data" / "filtered"

filtered_file = FILTERED_DIR / "fitness_filtered.jsonl"
raw_file = RAW_DIR / "fitness_train.jsonl"

with open(filtered_file, "r", encoding="utf-8") as f:
    passed_docs = [json.loads(line) for line in f]

with open(raw_file, "r", encoding="utf-8") as f:
    raw_docs = [json.loads(line) for line in f]

passed_texts = {doc["text"] for doc in passed_docs}
rejected_docs = [doc for doc in raw_docs if doc["text"] not in passed_texts]

print("=" * 80)
print("10 PASSED DOCUMENTS")
print("=" * 80)

for i, doc in enumerate(passed_docs[:10], 1):
    print(f"\n--- PASSED {i} ---")
    print(doc["text"])
    print(f"Characters: {len(doc['text'])}")

print("\n" + "=" * 80)
print("10 REJECTED DOCUMENTS")
print("=" * 80)

for i, doc in enumerate(rejected_docs[:10], 1):
    print(f"\n--- REJECTED {i} ---")
    print(doc["text"])
    print(f"Characters: {len(doc['text'])}")

print("\n" + "=" * 80)
print(f"Total Passed: {len(passed_docs)}")
print(f"Total Rejected: {len(rejected_docs)}")
print("=" * 80)
```

# 🔍 Filter tekshirish hisoboti

## 📌 Vazifa

10 ta **o‘tgan (passed)** va 10 ta **rad etilgan (rejected)** hujjat chop etilib, tekshirildi.

## ✅ O‘tgan hujjatlar

10 ta o‘tgan hujjatning barchasi **fitness va health** mavzusidagi normal savol-javoblardan iborat.

Ular foydali va dataset uchun mos.

## ❌ Rad etilgan hujjatlar

10 ta rad etilgan hujjat bir xil edi:

**Instruction: instruction**  
**Output: output**

Har birining uzunligi **39 ta belgi**.

Bu haqiqiy ma’lumot emas, balki **header yoki placeholder** ma’lumotga o‘xshaydi.

## 🔎 Qaysi filter ishladi?

Bu hujjatlar **`min_chars` filter** sababli rad etilgan!

Rad etilgan hujjatning uzunligi **39 characters**.

Aslida `min_chars` filter:

**`min_chars = 200`**

Hujjat uzunligi:

**39 < 200**

Shuning uchun bu hujjat **`min_chars` filterdan o‘tmadi** va rejected qilindi.

Boshqa filterlar bilan solishtirganda ham, bu hujjatning asosiy muammosi **juda qisqa ekanligi** bo‘ldi.

## 🎯 Xulosa

✅ **Passed** hujjatlar normal **fitness** savol-javoblar.

❌ **Rejected** hujjatlar esa bir xil **placeholder** ma’lumotlar.

📏 Ularning uzunligi **39 characters**, `min_chars` esa **200** bo'lishi kerak.

➡️ **39 < 200**, shuning uchun ular **`min_chars` filterdan o‘tmagan**.

💡 Demak, rejected hujjatlarning asosiy sababi **matn juda qisqa bo‘lgani**.
